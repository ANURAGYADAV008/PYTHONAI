```python
import chromadb
from youtube_transcript_api import YouTubeTranscriptApi
from youtube_transcript_api.formatters import TextFormatter
from langchain_ollama import ChatOllama
from chromadb.utils import embedding_functions
from sentence_transformers import SentenceTransformer
```

```python
chroma_client=chromadb.PersistentClient(path="my-vectordb")
```

```python
sentence_transformer_of=embedding_functions.SentenceTransformerEmbeddingFunction(model_name="sentence-transformers/all-mpnet-base-v2")
```

```python
chroma_collection=chroma_client.get_or_create_collection(name="yt_notes",embedding_function=sentence_transformer_of)
```

https://www.youtube.com/watch?v=H-oCV5brtU4

```python
yt_video_id='H-oCV5brtU4'
prompt="Extract key Notes from video transcript"
```

```python
transcript_list = YouTubeTranscriptApi().fetch(
    yt_video_id,
    languages=['en', 'en-US', 'en-GB']
)
transcript = TextFormatter().format_transcript(transcript_list)
```

```python
for list in transcript_list:
    print(list)
```

```python
with open("temp_transcript.txt","w") as file:
    file.write(transcript)
```

```python
llm=ChatOllama(model="qwen2.5-coder:3b")
res=llm.invoke("hello")
res
```

```python
res = llm.invoke(prompt + transcript, stream=False)
```

```python
print(res.content)
```

```python
with open("temp_notes.text",'w')as file:
    file.write(res.content)
```

```python
with open("temp_notes.txt")as file:
    notes=file.read()
    
chroma_collection.upsert(
    documents=[notes],
    ids=[yt_video_id],
    metadatas=[{
        "video_id": yt_video_id,
        "youtube_url": f"https://www.youtube.com/watch?v={yt_video_id}"
    }]
)

result=chroma_collection.get(yt_video_id,include=['documents','metadatas','embeddings'])
print(result)
```

```python
result = chroma_collection.query(
    query_texts=["fine tuning","what is FineTuning"],
    n_results=1,
    include=["metadatas",'documents']
)
print(result)

print(result["metadatas"][0][0]["youtube_url"])
```
