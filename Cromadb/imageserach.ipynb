```python
from chromadb.utils.embedding_functions import OpenCLIPEmbeddingFunction
from chromadb.utils.data_loaders import ImageLoader
```

```python
croma_client=chromadb.PersistentClient(path="my_vectordb")
```

```python
multimodel_of=OpenCLIPEmbeddingFunction()
```

```python
multimodel_db=croma_client.get_or_create_collection(name="multimodel",embedding_function=multimodel_of,data_loader=ImageLoader)
```

```python
from PIL import Image
import numpy as np

img1 = np.array(
    Image.open("C:\\Users\\jaysi\\Desktop\\Desktop\\PYTHONAI\\images\\pexels-213167965-17648457.jpg")
)
tiger="C:\\Users\\jaysi\\Desktop\\Desktop\\PYTHONAI\\images\\pexels-213167965-17648457.jpg"
lion="C:\\Users\\jaysi\Desktop\\Desktop\\PYTHONAI\\images\\pexels-luke-seago-20457134-11838539.jpg"
img2 = np.array(
    Image.open("C:\\Users\\jaysi\Desktop\\Desktop\\PYTHONAI\\images\\pexels-luke-seago-20457134-11838539.jpg")
)
```

```python
multimodel_db.delete(ids=["0", "1"])
```

```python
res=multimodel_db.add(
    ids=['0','1'],
    images=[img1,img2],
    metadatas=[
        {"file":"tiger"},
        {"file":"lion"}
    ]
)

print(multimodel_db.count())
multimodel_db.query(
    query_texts=['tiger'],
    n_results=2,
    include=['documents','metadatas','distances','data','embeddings']
)
```

```python
results = multimodel_db.query(
    query_texts=["tiger"],
    n_results=2,
    include=["metadatas", "distances"]
)
print(results)
for meta, dist in zip(
    results["metadatas"][0],
    results["distances"][0]
):
    print(meta["file"], dist)
```
